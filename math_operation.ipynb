# TENSOR

*   ADD,SUBTRACT,DIVIDE,MULTIPLY,REMAINDERS,EXPONENT

*   SHORTHAND AND LONGHAND

```python
import torch
import numpy as np
```

```python
tensor_a =torch.tensor([1,2,3,4])
tensor_b =torch.tensor([5,6,7,8])
```

```python
# addition
tensor_a + tensor_b
```

```python
## addition Longhand
torch.add(tensor_a , tensor_b )
```

```python
#subtract
tensor_b - tensor_a
```

```python
#subb function
torch.sub(tensor_b , tensor_a)
```

```python
#multiplication longhand
torch.mul(tensor_a , tensor_b)
```

```python
#divison longhand
torch.div(tensor_b , tensor_a)
```

```python
# remainder function
tensor_b % tensor_a
```

```python
# remainder longhand
torch.remainder(tensor_b , tensor_a)
```

```python
#tensor exponents
torch.pow(tensor_a,tensor_b)
```

```python
#another way for longhand
tensor_a.add(tensor_b)
```
